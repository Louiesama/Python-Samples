```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
import numpy as np
```

```python
df = pd.read_excel(r'C:\Users\99617\Desktop\lotto_result.xlsx')
df.head()
```

```python
winning_cols = ['n1','n2','n3','n4','n5','n6']
all_numbers = df[winning_cols].values.flatten()
```

```python
X = df[winning_cols].iloc[:-1].values
y = df[winning_cols].iloc[1:].values

model = RandomForestRegressor(n_estimators=100)
model.fit(X,y)

last_draw = df[winning_cols].iloc[-1].values.reshape(1,-1)
ml_prediction = model.predict(last_draw)
print(f"ML Predicted Numbers: {np.round(ml_prediction).astype(int)[0]}")
```
